# Section 1 - Working with Delimited Files

Delimited files, such as CSV (Comma-Separated Values), are a common way
to store structured data. Each value in the file is separated by a
specific character, often a comma. Julia’s DelimitedFiles package makes
it easy to work with these files.

> **Tip**
>
> The DelimitedFiles package is part of Julia’s Standard Library, which
> means you can use it without installing anything extra!

````julia
using DelimitedFiles
````

Now, let’s create a simple matrix and save it as a CSV file:

````julia
# Create a 3x3 matrix
new_data = [10 12 6; 13 25 1; 40 30 7]

# Create a new folder called "ExampleData"
# We use mkpath() instead of mkdir(), as mkpath() does nothing
# if the folder already exists - so you can safely re-run this block
mkpath("$(@__DIR__)/ExampleData")

# Write the matrix to a CSV file
open("$(@__DIR__)/ExampleData/matrix.csv", "w") do io
    writedlm(io, new_data, ',')
end

println("CSV file 'matrix.csv' written successfully to folder ExampleData!")
````

> **Note**
>
> Note that we used the `@__DIR__` macro to get the directory of the
> current file. This is a convenient way to ensure that the file path is
> correct, no matter where you run the script from. The reason is that
> the `@__DIR__` macro returns the directory of the file containing the
> macro, not your current working directory.

## Exercise 2.3 - Read the Brewery’s Order File

The order file of a brewery lands on your desk. The sales team exported
it from a German spreadsheet: columns are separated by `;`, and numbers
use a decimal comma, like `60,0`. The following code writes the file to
`ExampleData/orders.csv` exactly as you received it:

````julia
mkpath("$(@__DIR__)/ExampleData")
write("$(@__DIR__)/ExampleData/orders.csv", """
week;crates;note
1;53;
2;60,0;
3;61;
4;;
5;56;
6;126;harbour festival
7;60;
7;60;
8;59;
""")
println("File 'orders.csv' written to the folder ExampleData!")
````

First, read the file with plain `CSV.read(path, DataFrame)` and print
it. What type does the `crates` column have, and why? Then read the file
again into the variable `orders`, this time telling `CSV.read()` how the
file was written: `delim=';'` and `decimal=','`.

> **Tip**
>
> Keyword arguments come after a semicolon:
> `CSV.read(path, DataFrame; delim=';', decimal=',')`. Other tools treat
> the comma as a thousands separator and read `60,0` as 600.

````julia
# YOUR CODE BELOW
orders_path = "$(@__DIR__)/ExampleData/orders.csv"
# Read plainly, the crates column is text: "60,0" is not a number for CSV.read()
println(CSV.read(orders_path, DataFrame))
orders = CSV.read(orders_path, DataFrame; delim=';', decimal=',')
````

````julia
# Test your answer
@assert orders.crates[2] == 60.0 "Week 2 should read 60.0 crates. Did you set decimal=','?"
@assert ismissing(orders.crates[4]) "Week 4 has no number in the file, so it should be missing."
println("Order file read successfully!")
````

## Exercise 2.4 - Clean the Order File

Two problems remain: week 7 appears twice, and week 4 is `missing`.
Remove the duplicated row with `unique!()`, then replace the missing
value with the median of the other weeks. Week 6 looks unusual too, but
read its `note` before you touch it.

> **Tip**
>
> `median()` comes with the `Statistics` package (`using Statistics`; it
> ships with Julia). `skipmissing()` ignores missing values, and
> `coalesce.(column, value)` replaces every missing entry with `value`.

````julia
# YOUR CODE BELOW
using Statistics
unique!(orders)
# Week 6 stays: its note says the harbour festival was real
orders.crates = coalesce.(orders.crates, median(skipmissing(orders.crates)))
````

````julia
# Test your answer
@assert nrow(orders) == 8 "Week 7 should appear only once. Did you remove the duplicated row?"
@assert count(ismissing, orders.crates) == 0 "There are still missing values in the crates column."
@assert orders.crates[4] == 60.0 "Week 4 should get the median of the other weeks, 60 crates."
@assert orders.crates[6] == 126 "The harbour festival in week 6 was real: keep its 126 crates."
println("Order file cleaned successfully!")
````
